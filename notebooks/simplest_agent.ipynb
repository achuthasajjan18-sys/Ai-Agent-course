# Creating a Simple Agent with Tracing

```python
import dotenv
import os

from openai import OpenAI

dotenv.load_dotenv()

if not os.environ.get("OPENAI_API_KEY"):
    print(
        "OPENAI_API_KEY is not set. Add it to your environment to run the agent."
    )
else:
    print(
        OpenAI()
        .responses.create(
            model=os.environ.get("OPENAI_DEFAULT_MODEL", "gpt-4o-mini"),
            input="Say: We are up and running!",
        )
        .output_text
    )
```

```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

Create a Placement Trainer Agent

```python
placement_trainer_agent = Agent(
    name="Placement Trainer",
    instructions="""
    You are a helpful placement preparation guide for students.
    Give concise, practical answers in clear language.
    Help with aptitude, reasoning, coding, data structures, technical interviews,
    HR interviews, resumes, communication, and placement strategies.
    When useful, provide a short step-by-step plan, an example, or a practice question.
    Ask one brief clarifying question only when the user's goal is unclear.
    Do not invent company-specific hiring details; say when information may vary.
    """,
)
```

Let's execute the Placement Trainer:

```python
if not os.environ.get("OPENAI_API_KEY"):
    print("Set OPENAI_API_KEY to run the Placement Trainer.")
else:
    with trace("Placement Trainer Agent"):
        result = await Runner.run(
            placement_trainer_agent,
            "How should I prepare for a software engineering placement in 30 days?",
        )

    print(result.final_output)
```

Streaming the placement advice token by token

```python
if not os.environ.get("OPENAI_API_KEY"):
    print("Set OPENAI_API_KEY to stream placement advice.")
else:
    response_stream = Runner.run_streamed(
        placement_trainer_agent,
        "Give me three focused topics to practice for a coding placement test.",
    )

    async for event in response_stream.stream_events():
        if event.type == "raw_response_event" and isinstance(
            event.data, ResponseTextDeltaEvent
        ):
            print(event.data.delta, end="", flush=True)
```
